```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread("images/ope9/image5.png", cv2.IMREAD_GRAYSCALE)

if image is None : 
    raise FileNotFoundError("unable to load image")

print(image.shape)

plt.figure(figsize=(24,16))

plt.subplot(2,3,1)
plt.title("Original Image")
plt.axis("off")
plt.imshow(image,cmap="gray")

# sharpening kernel
sharpening_kernel = np.array(
    [[-1,-1,-1],
    [-1,9,-1],
    [-1,-1,-1]]
)
sharpened_img = cv2.filter2D(image,-1,sharpening_kernel)

plt.subplot(2,3,2)
plt.imshow(sharpened_img,cmap="gray")
plt.title("sharpened Image")
plt.axis("off")

# Averaging linear filter

kernel_size = (5,5)

kernel = np.ones(kernel_size,dtype=np.float32)/(kernel_size[0]*kernel_size[1])

smoothened_avg_img = cv2.filter2D(image,-1,kernel)

plt.subplot(2,3,3)
plt.title("After applying Averaging linear filter")
plt.axis("off")
plt.imshow(smoothened_avg_img,cmap="gray")


# median filters

kernel_size = 5

smoothened_med_img = cv2.medianBlur(image,kernel_size)

plt.subplot(2,3,4)
plt.title("After applying median filters")
plt.axis("off")
plt.imshow(smoothened_med_img,cmap="gray")

# Gaussian Blur 

blurred_img = cv2.GaussianBlur(image,(11,11),0)

plt.subplot(2,3,5)
plt.imshow(blurred_img,cmap="gray")
plt.title("blurred Image")
plt.axis("off")

# sharpening using spatial high pass filters

sharpening_kernel = np.array(
    [[0,-1,0],
     [-1,5,-1],
     [0,-1,0]],dtype=np.float32
)

sharpened_img = cv2.filter2D(image, -1, sharpening_kernel)

plt.subplot(2,3,6)
plt.imshow(sharpened_img,cmap="gray")
plt.title("Sharpened Image")
plt.axis("off")

plt.tight_layout()

plt.show()
```

```python
# good result came from median filtering , then we can apply simple sharpening
import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread("images/ope9/image1.png", cv2.IMREAD_GRAYSCALE)

if image is None : 
    raise FileNotFoundError("unable to load image")

plt.figure(figsize=(12,8))


plt.subplot(2,2,1)
plt.title("Original Image")
plt.axis("off")
plt.imshow(image,cmap="gray")

# apply median filtering for removing salt and paper noise

kernel_size = 5

smoothened_med_img = cv2.medianBlur(image,kernel_size)

plt.subplot(2,2,2)
plt.title("After applying median filters")
plt.axis("off")
plt.imshow(smoothened_med_img,cmap="gray")

# sharpening using spatial high pass filters

sharpening_kernel = np.array(
    [[-1,-1,-1],
     [-1,9,-1],
     [-1,-1,-1]],dtype=np.float32
)

full_sharpened = cv2.filter2D(image, -1, sharpening_kernel)

plt.subplot(2,2,3)
plt.imshow(full_sharpened,cmap="gray")
plt.title("full sharpened Image")
plt.axis("off")

sharpened_img = (
    smoothened_med_img * 0.85 +
    full_sharpened * 0.15
).astype(np.uint8)

plt.subplot(2,2,4)
plt.imshow(sharpened_img,cmap="gray")
plt.title("Sharpened Image")
plt.axis("off")

plt.tight_layout()

plt.show()
```
