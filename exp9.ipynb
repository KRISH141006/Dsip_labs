```python
# write a python program to simulate smoothening and sharpening operation on image using spatial filter

import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread("images/mountain.jpg")

if image is None : 
    raise FileNotFoundError("unable to load image")

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

kernel_size = (5,5)
sigma = 1.5

gaussian_kernel = cv2.getGaussianKernel(kernel_size[0],sigma)

gaussian_kernel = np.outer(gaussian_kernel,gaussian_kernel)

smoothened_img = cv2.filter2D(image,-1,gaussian_kernel)

sharpening_kernel = np.array(
    [[-1,-1,-1],
    [-1,9,-1],
    [-1,-1,-1]]
)
sharpened_img = cv2.filter2D(image,-1,sharpening_kernel)

plt.figure(figsize=(16,8))

plt.subplot(1,3,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(smoothened_img)
plt.title("Smoothened Image")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(sharpened_img)
plt.title("Sharpened Image")
plt.axis("off")

plt.tight_layout()

plt.show()
```

```python
# write a python program to simulate smoothening using Averaging linear filters

import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread("images/mountain.jpg")

if image is None : 
    raise FileNotFoundError("unable to load image")

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

kernel_size = (5,5)

kernel = np.ones(kernel_size,dtype=np.float32)/(kernel_size[0]*kernel_size[1])

smoothened_avg_img = cv2.filter2D(image,-1,kernel)

plt.figure(figsize=(16,8))

plt.subplot(1,2,1)
plt.title("Original Image")
plt.axis("off")
plt.imshow(image)

plt.subplot(1,2,2)
plt.title("After applying Averaging linear filter")
plt.axis("off")
plt.imshow(smoothened_avg_img)

plt.tight_layout()

plt.show()
```

```python
# write a python program to simulate smoothening using median filters

import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread("images/mountain.jpg")

if image is None : 
    raise FileNotFoundError("unable to load image")

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

kernel_size = 11

smoothened_med_img = cv2.medianBlur(image,kernel_size)

plt.figure(figsize=(16,8))

plt.subplot(1,2,1)
plt.title("Original Image")
plt.axis("off")
plt.imshow(image)

plt.subplot(1,2,2)
plt.title("After applying median filters")
plt.axis("off")
plt.imshow(smoothened_med_img)

plt.tight_layout()

plt.show()
```

```python
# write a python program to simulate sharpening using spatial high pass filters

import cv2
import matplotlib.pyplot as plt
import numpy as np

image = cv2.imread("images/mountain.jpg")

if image is None : 
    raise FileNotFoundError("unable to load image")

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

blurred_img = cv2.GaussianBlur(image,(11,11),0)

sharpening_kernel = np.array(
    [[0,-1,0],
     [-1,5,-1],
     [0,-1,0]],dtype=np.float32
)

sharpened_img = cv2.filter2D(blurred_img, -1, sharpening_kernel)

plt.figure(figsize=(16,8))

plt.subplot(1,3,1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(blurred_img)
plt.title("blurred Image")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(sharpened_img)
plt.title("Sharpened Image")
plt.axis("off")

plt.tight_layout()

plt.show()
```
