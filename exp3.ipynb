```python
import numpy as np
import matplotlib.pyplot as plt

signal1 = np.array([1,2,3,4,5])
signal2 = np.array([6,7,8,9,10])

lags = np.arange(-4,5)

# finding autocorrelation
autocorr = np.correlate(signal1,signal1,mode='full')

# finding crosscorrelation
crosscorr = np.correlate(signal1,signal2,mode='full')

# plotting both
plt.figure(figsize=(10,6))

plt.subplot(2,1,1)
plt.stem(lags,autocorr)
plt.title("Autocorrelation")
plt.xlabel("Lag")
plt.ylabel("Correlation")
plt.grid()
plt.subplot(2,1,2)
plt.stem(lags,crosscorr)
plt.title("Crosscorrelation")
plt.xlabel("Lag")
plt.grid()
plt.ylabel("Correlation")

plt.tight_layout()

plt.show()
```
