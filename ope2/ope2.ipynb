```python
from pydub import AudioSegment
import numpy as np
from scipy.signal import convolve
import matplotlib.pyplot as plt

audio = AudioSegment.from_wav("audios/mixed_audio.wav")

audio = audio.set_channels(1)

samples = np.array(audio.get_array_of_samples()).astype(np.float32)
```

```python
# impulse responce 1
h1 = np.array([0.2,0.6,0.2])

filtered1 = convolve(samples, h1, mode="same")

filtered1 = filtered1 / np.max(np.abs(filtered1))

filtered1_int16 = (filtered1 * 32767).astype(np.int16)

output_audio = AudioSegment(
    filtered1_int16.tobytes(),
    frame_rate=audio.frame_rate,
    sample_width=2,
    channels=1
)

output_audio.export("audios/output_h1.wav", format="wav")

# Number of samples to display
num_samples = 20000

plt.figure(figsize=(15, 6))

# Original mixed audio
plt.subplot(2, 1, 1)
plt.plot(samples[:num_samples])
plt.title("Original Mixed Audio")
plt.xlabel("Sample Number")
plt.ylabel("Amplitude")

# Processed audio
plt.subplot(2, 1, 2)
plt.plot(filtered1_int16[:num_samples])
plt.title("Processed Audio")
plt.xlabel("Sample Number")
plt.ylabel("Amplitude")

plt.tight_layout()
plt.show()
```
