```python
import numpy as np
import matplotlib.pyplot as plt

signal = np.array([1, 2, 3, 4, 5, 6, 7, 8])

# Compute the FFT of the signal
fft_result = np.fft.fft(signal)

# Compute the magnitude and Phase spectrum of the FFT result
magnitude_spectrum = np.abs(fft_result)
Phase_spectrum = np.angle(fft_result)

# Compute the IFFT of the FFT result
reconstructed_signal = np.fft.ifft(fft_result)

# Display the results
print("Original Signal      :", signal)
print("FFT Result           :", np.round(fft_result, 4))
print("Magnitude Spectrum   :", np.round(magnitude_spectrum, 4))
print("Phase Spectrum       :", np.round(Phase_spectrum, 4))
print("Reconstructed Signal :", np.round(reconstructed_signal.real, 4))

# Plotting
plt.figure(figsize=(10, 8))

plt.subplot(2, 2, 1)
plt.stem(signal)
plt.title("Original Signal")
plt.xlabel("n")
plt.ylabel("Amplitude")

plt.subplot(2, 2, 2)
plt.stem(magnitude_spectrum)
plt.title("Magnitude Spectrum")
plt.xlabel("k")
plt.ylabel("|X(k)|")

plt.subplot(2, 2, 3)
plt.stem(Phase_spectrum)
plt.title("Phase Spectrum")
plt.xlabel("k")
plt.ylabel("Phase (radians)")

plt.subplot(2, 2, 4)
plt.stem(reconstructed_signal.real)
plt.title("Reconstructed Signal (IFFT)")
plt.xlabel("n")
plt.ylabel("Amplitude")

plt.tight_layout()
plt.show()
```
