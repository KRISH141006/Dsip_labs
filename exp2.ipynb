```python
import numpy as np
import matplotlib.pyplot as plt

signal1 = np.array([1,2,3,4,5])
signal2 = np.array([6,7,8,9,10])

# finding linear convolution
linear_conv = np.convolve(signal1,signal2,mode='full')


# finding circular convolution by converting the signal into frequency domain and multiply both the signals and applying inverse fft
fft_len = max(len(signal1),len(signal2))
signal1_fft = np.fft.fft(signal1,fft_len)
signal2_fft = np.fft.fft(signal2,fft_len)

circular_conv = np.fft.ifft(signal1_fft*signal2_fft).real

# plotting both
plt.figure(figsize=(10,6))

plt.subplot(2,1,1)
plt.stem(linear_conv)
plt.title("Linear Convolution")
plt.xlabel("Sample")
plt.ylabel("Amplitude")

plt.subplot(2,1,2)
plt.stem(circular_conv)
plt.title("circular Convolution")
plt.xlabel("Sample")
plt.ylabel("Amplitude")

plt.tight_layout()

plt.show()
```
