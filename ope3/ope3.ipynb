```python
from pydub import AudioSegment
import numpy as np

music1 = AudioSegment.from_mp3("audio/karaoke_Everything_I_Do.mp3")
song1 = AudioSegment.from_mp3("audio/song_Everything_I_Do.mp3")
song2 = AudioSegment.from_mp3("audio/kaun_tujhe.mp3")

music = music1.set_channels(1)
song1 = song1.set_channels(1)
song2 = song2.set_channels(1)

music1_samples = np.array(music1.get_array_of_samples()).astype(np.float32)
song1_samples = np.array(song1.get_array_of_samples()).astype(np.float32)
song2_samples = np.array(song2.get_array_of_samples()).astype(np.float32)

min_length = min(len(music1_samples),len(song1_samples),len(song2_samples))
```

```python
# correlation

music1_samples = music1_samples[:min_length]
song1_samples = song1_samples[:min_length]
song2_samples = song2_samples[:min_length]

# correlation between song1 and music1
corr_original_karaoke = np.corrcoef(song1_samples, music1_samples)[0, 1]

# correlation between song1 and song2
corr_original_different = np.corrcoef(song1_samples, song2_samples)[0, 1]

# correlation between music1 and song2
corr_karaoke_different = np.corrcoef(music1_samples, song2_samples)[0, 1]

print("Original vs Karaoke:", corr_original_karaoke)
print("Original vs Different:", corr_original_different)
print("Karaoke vs Different:", corr_karaoke_different)
```

```python
# cross - correlation
from scipy.signal import correlate

# Normalize signals
song1_norm = song1_samples / np.max(np.abs(song1_samples))
music1_norm = music1_samples / np.max(np.abs(music1_samples))
song2_norm = song2_samples / np.max(np.abs(song2_samples))

# Cross-correlation
cross_corr_org_krk = correlate(song1_norm, music1_norm, mode="full")
cross_corr_org_diff = correlate(song1_norm, song2_norm, mode="full")
cross_corr_krk_diff = correlate(song2_norm, music1_norm, mode="full")

# Normalize the correlation
cross_corr_org_krk = cross_corr_org_krk / (
    np.linalg.norm(song1_norm) * np.linalg.norm(music1_norm)
)
cross_corr_org_diff = cross_corr_org_diff / (
    np.linalg.norm(song1_norm) * np.linalg.norm(song2_norm)
)
cross_corr_krk_diff = cross_corr_krk_diff / (
    np.linalg.norm(song2_norm) * np.linalg.norm(music1_norm)
)

# Find highest correlation
max_corr_org_krk = np.max(np.abs(cross_corr_org_krk))
max_corr_org_diff = np.max(np.abs(cross_corr_org_diff))
max_corr_krk_diff = np.max(np.abs(cross_corr_krk_diff))

print("Maximum cross-correlation of org & krk:", max_corr_org_krk)
print("Maximum cross-correlation of org & diff:", max_corr_org_diff)
print("Maximum cross-correlation of krk & diff:", max_corr_krk_diff)
```
