```python
# 1. Unit impulse Sequence

import numpy as np
import matplotlib.pyplot as plt

x=np.arange(-10,11)
unit_impulse = (x == 0).astype(int)

plt.stem(x, unit_impulse)
# plt.figure(figsize=(8,5))
plt.title("Unit impulse Sequence")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid()
plt.show()
```

```python
#2. Unit Step Sequence

x=np.arange(-10,11)
unit_step = (x >= 0).astype(int)

plt.stem(x,unit_step)
plt.title("Unit Step Sequence")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid()
plt.show()
```

```python
#3. Unit Ramp Sequence

x=np.arange(-10,11)
# unit_ramp = [i if i>=0 else 0 for i in x]
unit_ramp = np.maximum(x, 0)

plt.stem(x,unit_ramp)
plt.title("Unit Ramp Sequence")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid()
plt.show()
```

```python
#4. Exponential Sequence

x = np.arange(-5, 6)
a = 1.1
exp_ = a**x


plt.stem(x,exp_)
plt.title("Exponential Sequence")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
#5. Sinusoidal Sequence

x = np.arange(-10,11)
A = 1
w = np.pi/4

sinusoidal = A*np.sin(x*w)

plt.plot(x,sinusoidal,color="red")
plt.title("Sinusoidal Sequence")
plt.xlabel("n")
plt.ylabel("Amplitude")
plt.grid()
plt.show()
```
