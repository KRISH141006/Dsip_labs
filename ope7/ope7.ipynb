```python
# 1

import cv2
import numpy as np
import matplotlib.pyplot as plt

src_img = cv2.imread("images/image2.png",cv2.IMREAD_GRAYSCALE)

if src_img is None:
    raise FileNotFoundError("Unable to load image")

# Converting it into negative image
neg_img = 255 - src_img

# Thresholding image
_,thres_img = cv2.threshold(src_img,128,255,cv2.THRESH_BINARY)

# Gamma correction
norm_img = src_img/255.0
gamma = 1.5
raising_power = np.power(norm_img,1/gamma)
gamma_img = np.uint8(raising_power * 255)

plt.figure(figsize=(6,4))

plt.subplot(2,2,1)
plt.title("Original Image")
plt.axis("off")
plt.imshow(src_img,cmap="gray")

plt.subplot(2,2,2)
plt.title("Negative Image")
plt.axis("off")
plt.imshow(neg_img,cmap="gray")

plt.subplot(2,2,3)
plt.title(" Image after Thresholding")
plt.axis("off")
plt.imshow(thres_img,cmap="gray")

plt.subplot(2,2,4)
plt.title("Gamma Image")
plt.axis("off")
plt.imshow(gamma_img,cmap="gray")

plt.tight_layout()

plt.show()
```
